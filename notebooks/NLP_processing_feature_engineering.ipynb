```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot
import sklearn
import joblib
```

```python
data = joblib.load("../artifacts/processed_data.joblib")
```

```python
data.head()
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
data["tags"][0]
```

```python
def stemming(text):
    L = []
    for i in text.split():
        L.append(ps.stem(i))
    return " ".join(L)
```

```python
data['tags']=data['tags'].apply(stemming)
```

```python
data["tags"]
```

Vectorization


```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(max_features= 5000, stop_words='english')
```

```python
vectors= cv.fit_transform(data["tags"])
```

```python
vectors = vectors.toarray()
```

```python
vectors
```

```python
cv.get_feature_names_out()[:200]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity
```

```python
similarity [0]
```

```python
data[data["title"]=='avatar'].index[0]
```

```python
def recommend(movie):

    movie_index= data[data["title"]== movie].index[0]

    recommendations = similarity[movie_index]

    movie_list= sorted(enumerate(recommendations),reverse=True, key=lambda x: x[1])[1:6]
    
    for i in movie_list:
        print(data.iloc[i[0]].title)
        
```

```python
import joblib
```

```python
joblib.dump( similarity,
"../artifacts/similarity.joblib", compress=5)
```

```python
recommend('avatar')
```
